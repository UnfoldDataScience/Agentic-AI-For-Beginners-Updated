## Before you start (AWS prerequisites)

1. An AWS account. Bedrock calls are **billed to your account**.
2. The AWS CLI installed and signed in with **your own** IAM user: run `aws configure`.
   Give that user only the permissions it needs (for example the `AmazonBedrockFullAccess` managed policy)
   rather than administrator access.
3. In the Bedrock console, enable access to the model you want to use, in your region.
4. Put your settings in the course `.env` file (see `.env.example`). Nothing account-specific is written in this notebook.

> **Compatibility update (differs from the recording)**
> - The model is called with Bedrock's **Converse API** instead of `invoke_model` with a Titan-specific JSON body.
>   The default model is still Amazon Titan Text Express. To use another Bedrock text model, set `BEDROCK_MODEL_ID` in `.env`; no code changes are needed.
> - The AWS region and model come from `.env` instead of being written in the code.

```python
import os

import boto3
from dotenv import load_dotenv

load_dotenv()

AWS_REGION = os.getenv("AWS_REGION") or "us-east-1"
MODEL_ID = os.getenv("BEDROCK_MODEL_ID") or "amazon.titan-text-express-v1"
print("Region:", AWS_REGION, "| Model:", MODEL_ID)
```

```python
bedrock_runtime = boto3.client("bedrock-runtime", region_name=AWS_REGION)
```

```python
prompt = "Hi tell me about making a good PPT for marketing"

kwargs = {
    "modelId": MODEL_ID,
    "messages": [{"role": "user", "content": [{"text": prompt}]}],
    "inferenceConfig": {
        "maxTokens": 512,
        "temperature": 0,
        "topP": 0.9
    }
}
```

```python
response = bedrock_runtime.converse(**kwargs)
```

```python
generation = response["output"]["message"]["content"][0]["text"]
```

```python
print(generation)
```
